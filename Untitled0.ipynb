```python
from transformers import pipeline
import gradio as gr

# Ye chota aur tez model hai, jaldi load hoga
classifier = pipeline("sentiment-analysis", model="cardiffnlp/twitter-roberta-base-sentiment-latest")

def analyze(text):
    result = classifier(text)[0]
    label = result['label']
    score = round(result['score']*100, 2)
    return f"Result: {label} ({score}%)"

# Ye interface banayega
demo = gr.Interface(
    fn=analyze,
    inputs=gr.Textbox(label="Apna Jumla Likho / Enter Text", placeholder="I love Pakistan..."),
    outputs=gr.Textbox(label="Result"),
    title="AI Sentiment Analyzer - Agentic AI Project",
    description="Ye project Hugging Face Transformers par bana hai. Positive / Negative / Neutral detect karta hai."
)

demo.launch()/
```
